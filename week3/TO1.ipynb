```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

gates = ['x', 'y', 'z']

for gate in gates:
    qc = QuantumCircuit(1)

    if gate == 'x':
        qc.x(0)
    elif gate == 'y':
        qc.y(0)
    else:
        qc.z(0)

    state = Statevector.from_instruction(qc)

    print(f"\n----- {gate.upper()} Gate -----")
    print(qc)
    print(state)
```
