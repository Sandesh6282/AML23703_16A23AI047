```python
import random
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

gate_list = ['x','y','z','h','s','t']

qc = QuantumCircuit(1)

selected = random.sample(gate_list,5)

print("Selected Gates:", selected)

for gate in selected:
    getattr(qc, gate)(0)

state = Statevector.from_instruction(qc)

print(qc)
print("\nFinal State:")
print(state)
```
