```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import matplotlib.pyplot as plt
```

```python
shots = 1024

one_qubit = QuantumCircuit(1,1)

one_qubit.h(0)

one_qubit.measure(0,0)

print(one_qubit.draw())
```

```python
sim = AerSimulator()

result = sim.run(one_qubit, shots=shots).result()

measurement = result.get_counts()

print("Measurement Result:")
print(measurement)
```

```python
states = list(measurement.keys())

values = list(measurement.values())


plt.figure(figsize=(6,6))

plt.pie(values,
        labels=states,
        autopct='%1.2f%%',
        startangle=90)

plt.title("1 Qubit Hadamard Probability Distribution")

plt.show()
```

```python
print("Probabilities")

for state,count in measurement.items():

    print(state,"=",round(count/shots,4))
```
