```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import random
import matplotlib.pyplot as plt
```

```python
shots = 2000


random_circuit = QuantumCircuit(1,1)


random_circuit.h(0)

random_circuit.measure(0,0)


simulator = AerSimulator()


quantum_output = simulator.run(
    random_circuit,
    shots=shots
).result().get_counts()


print(quantum_output)
```

```python
classical_output = {

    "0":0,

    "1":0
}


for i in range(shots):

    bit=str(random.randint(0,1))

    classical_output[bit]+=1



print(classical_output)
```

```python
labels=["0","1"]


quantum_values=[
    quantum_output.get(x,0)
    for x in labels
]


python_values=[
    classical_output[x]
    for x in labels
]
```

```python
plt.figure(figsize=(7,5))


plt.scatter(
    labels,
    quantum_values,
    s=150,
    label="Quantum"
)


plt.scatter(
    labels,
    python_values,
    s=150,
    label="Python"
)


plt.xlabel("Generated Bit")

plt.ylabel("Frequency")


plt.title("Randomness Comparison")


plt.legend()

plt.grid(True)


plt.show()
```

```python
print("Generator     0 Probability     1 Probability")


print(
"Quantum       ",
round(quantum_values[0]/shots,3),
round(quantum_values[1]/shots,3)
)


print(
"Python        ",
round(python_values[0]/shots,3),
round(python_values[1]/shots,3)
)
```
