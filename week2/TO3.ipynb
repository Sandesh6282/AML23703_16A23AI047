```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
import numpy as np

# Create two-qubit circuit
qc = QuantumCircuit(2)

# Apply different operations
qc.h(1)
qc.cx(1, 0)

# Generate statevector
state = Statevector.from_instruction(qc)

print("Circuit:")
print(qc)

print("\nStatevector:")
print(state)

print("\nAmplitude of each basis state:")

for i, value in enumerate(state.data):
    print(f"|{format(i,'02b')}> -> {value}")

print("\nNormalization =", np.sum(np.abs(state.data)**2))
```
