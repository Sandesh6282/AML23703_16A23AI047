```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
import numpy as np

# Two-qubit circuit
qc = QuantumCircuit(2)

# Different gate combination
qc.y(0)
qc.h(1)
qc.cz(0, 1)

# Generate statevector
state = Statevector.from_instruction(qc)

print("Quantum Circuit:")
print(qc)

print("\nStatevector:")
print(state)

print("\nProbabilities:")

for i, amp in enumerate(state.data):
    probability = abs(amp)**2
    print(f"|{format(i,'02b')}> : {probability:.4f}")

print("\nTotal Probability =", np.sum(np.abs(state.data)**2))
```
