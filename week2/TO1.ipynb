```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Initialize one-qubit circuit
qc = QuantumCircuit(1)

# Apply Hadamard gate
qc.h(0)

# Obtain statevector
sv = Statevector.from_instruction(qc)

print("Circuit:")
print(qc)

print("\nGenerated Statevector:")
print(sv)
```
